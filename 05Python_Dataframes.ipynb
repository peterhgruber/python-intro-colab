### 6 Aggregation functions
* Use `df.xxx()` methods to aggregate/summarize the data
* Widely used are `mean()`, `median()`, `min()`, `max()`, `sum()`, `count()`, and `unique()`
* We have already seen `describe()`

```python
# Mean and median grade
print(student['math'].mean())
print(student['math'].median())
```

```python
# Unique values
student['isMale'].unique()
```

```python
# Aggregation can be applied to logical conditions
# Passing grade is "50% or more"
student['math'] >= 50
```

```python
# Fraction of students who pass mathematics
(student['math'] >= 50).mean()
```

### 7 Subsetting by condition

* Daily bread of data scientist. Analyze ...
    * "all male students"
    * "all male students who pass mathematics"
    * "all male students from the city who pass mathematics"
* **Syntax:** `df[<condition> ]`
* **Syntax:** `df['<column>'][<condition> ]`

```python
# "all male students"
student[ student['isMale']==1 ]
```

#### 7.1 Multiple conditions
- Requires specific syntax (`~`, `&`,  `|`) instead of  `not`, `and`, `or`.
- Also requires extra set of round brackets
  - **Syntax:** `df[ (<condition1>) & (<condition2>) ]`


```python
# "all male students who pass mathematics"
student[ (student['isMale']==1) & (student['math']>50) ]
```

```python
# "all male students from the city who pass mathematics"
student[ (student['isMale']==1) & (student['isCity']==1) & (student['math']>50) ]
```

#### *7.2 All possible subsets with `.groupby()`

* Interest is often in *all* combination of criteria, e.g. male/city, female/city, male/countryside and female/countrysde
* This can be done by `.groupby(<criterion>)`

```python
# Average math grade by gender
student.groupby('isMale')['math'].mean()
```

```python
# Average math grade by gender and city
student.groupby(['isMale','isCity'])['math'].mean()
```

## *9 Exporting to LaTeX

* Writing LaTeX tables can be tedious
* Pandas lets you export a DataFrame to the LaTeX table format
* **However,** often we don't want to export the data itself, but rather a statistical summary $\rightarrow$ this requires one extra step

```python
# Recall
student.describe()
```

```python
# To export the above output to LaTeX, we need to create
# a second DataFrame called student_summary
student_summary = student.describe()
student_summary
```

```python
# Now we can export student_summary to LaTeX
student_summary.to_latex('student.tex')
```

***Formatting:** the `float_format` parameter controls the number of digits.

For example `'%.2f'` means *"2 decimal places"*.

```python
student_summary.to_latex('student_2.tex', float_format='%.2f')
```

### *5.1 Multiple and complementary conditons .. not that easy

- Pandas is an abstraction that simplifies data operations, but...
    - Has different Object Types: Series and DataFrame
    - Requires specific syntax (`~`, `&`,  `|`) instead of  `not`, `and`, `or`.

```python
# A "DataFrame" object
print(type(student))

# Single column is seen as "Series" object
print(type(student['Age']))

# Multiple columns are seen as "DataFrame" object
print(type(student[ ['Age', 'Purchased'] ]))
```

```python
# Complementary conditon, "~" for "not"
~(student['Age'] <=40)
```

```python
# Dual conditions, "&" for "and"
# Users with age below 40 and salary above 50'000
(student['Age']<=40) & (student['EstimatedSalary']>=50000)
```

```python
# "|" for "or"
# Users with age either under 30 or above 45
(student['Age']<=30) | (student['Age']>=45)
```

```python
# Fraction of people above 40
( ~(student['Age']<=40) ).mean()
```

```python
# Tertia non datur
( ~(student['Age']<=40) ).mean() + (student['Age']<=40).mean()
```

### 6 Subsetting by condition

* Daily bread of data scientist. Analyze ...
    * "all older participants"
    * "all older male participants"
    * "all older male participants with high income"
* **Syntax:** `df[<condition> ]`
* **Syntax:** `df['<column>'][<condition> ]`

```python
# Recap
student['Age']>=30
```

```python
# Select rows for which <condtion> is True
student[ student['Age']>=30 ]
```

```python
# Subset just one column
student['EstimatedSalary'][ student['Age']>=30 ]
```

**Notice:** Length is now shorter

```python
# Statistics of a subset
# Average salary of people aged 30 or more.
student['EstimatedSalary'][ student['Age']>=30 ].mean()
```

#### 6.1 Multiple conditions
* Require extra `()` around condition
* Still `&` and `|` operators
* **Syntax** `df['<column>'][ (<condition1>) & (<condition2>) ]`

```python
student['EstimatedSalary'][ (student['Age']>=30) & (student['Gender']=="Male") ].mean()
```

```python
student['EstimatedSalary'][ (student['Age']>=30) & (student['Gender']=="Male") &
                       (student['EstimatedSalary']>40000) ].mean()
```

## *8 Exporting to LaTeX

* Writing LaTeX tables can be tedious
* Pandas lets you export a DataFrame to the LaTeX table format
* **However,** often we don't want to export the data itself, but rather a statistical summary $\rightarrow$ this requires one extra step

```python
# Recall
student.describe()
```

```python
# To export the above output to LaTeX, we need to create
# a second DataFrame called student_summary
student_summary = student.describe()
student_summary
```

```python
# Now we can export student_summary to LaTeX
student_summary.to_latex('student.tex')
```

***Formatting:** the `float_format` parameter controls the number of digits.

For example `'%.2f'` means *"2 decimal places"*.

```python
student_summary.to_latex('student_2.tex', float_format='%.2f')
```
